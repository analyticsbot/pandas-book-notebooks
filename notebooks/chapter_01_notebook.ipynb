## 1.1 What Is pandas?

pandas is an open-source Python library for working with structured, labeled data. The core idea: **data has labels** — rows have labels, columns have names, and every column can hold a different type.

Here is the simplest possible pandas code:

````python
df = pd.DataFrame({
    'name': ['Alice', 'Bob', 'Charlie'],
    'salary': [70000, 85000, 60000]
})

print(df)
````

Every row has a numeric label (0, 1, 2) and every column has a name. That labeling — invisible in a plain Python list — is what makes pandas useful.

---
## 1.2 Why pandas Exists — NumPy vs pandas

Before pandas, Python developers used NumPy arrays. NumPy is fast but has no labels.

````python
import numpy as np

# NumPy: fast, but no labels
data = np.array([[70000, 25], [85000, 30], [60000, 28]])
print("NumPy array:")
print(data)
print("Which column is salary? Which is age? You have to remember.")
print()

# pandas: labeled and clear
df = pd.DataFrame({
    'salary': [70000, 85000, 60000],
    'age': [25, 30, 28]
})
print("pandas DataFrame:")
print(df)
print("df['salary'] — no ambiguity.")
````

**Key insight:** NumPy gives you speed with unlabeled arrays. pandas adds the labels, mixed types, and missing-value handling that real data requires.

---
## 1.3 Core Data Structures (Preview)

pandas has two data structures: **Series** (1D) and **DataFrame** (2D). Chapter 3 goes deep — this is a preview.

### Series: One-Dimensional

A Series is a single column of labeled data.

````python
salaries = pd.Series([70000, 85000, 60000], name='salary')
print(salaries)
````

### DataFrame: Two-Dimensional

A DataFrame is a table — multiple columns sharing the same row labels. Think of it as a dictionary of Series.

````python
employees = pd.DataFrame({
    'name': ['Alice', 'Bob', 'Charlie', 'Diana'],
    'department': ['Sales', 'Engineering', 'Sales', 'HR'],
    'salary': [50000, 75000, 55000, 60000]
})

print(employees)
print(f"\nShape: {employees.shape}")
print(f"Columns: {list(employees.columns)}")
print(f"\nData types:\n{employees.dtypes}")
````

### Break This

What happens when rows have different lengths?

````python
broken = pd.DataFrame([[1, 2], [3]])
print(broken)
````

The second row has only one value, so pandas fills the gap with `NaN` (Not a Number). No error, no warning — just a silent `NaN`. This is why labeled, structured data matters.

You will learn to detect and handle `NaN` properly in Chapter 10.

## 1.4 pandas in the Python Data Ecosystem

pandas sits at the center of the Python data stack:

```
NumPy (arrays) -> pandas (tables) -> Matplotlib/Seaborn (plots) -> scikit-learn (models)
```

| Library | Role | How it connects to pandas |
|---|---|---|
| **NumPy** | Fast numerical arrays | pandas is built on NumPy. Every DataFrame column is a NumPy array under the hood. |
| **Matplotlib** | Low-level plotting | `df.plot()` calls Matplotlib. Chapter 18 covers this. |
| **Seaborn** | Statistical visualization | Takes pandas DataFrames directly as input. |
| **scikit-learn** | Machine learning | Expects data in DataFrame or array form. pandas prepares the data, scikit-learn models it. |
| **Polars** | High-performance DataFrames | An alternative to pandas for very large datasets. Uses a similar concept but different API. |
| **Dask** | Parallel computing | Extends pandas to datasets that do not fit in memory. |
| **cuDF** | GPU-accelerated DataFrames | Similar API to pandas but runs on NVIDIA GPUs for faster computation. |

---
## 1.5 When to Use pandas (and When Not To)

| Scenario | Best tool | Why |
|---|---|---|
| Tabular data, <10M rows, exploratory analysis | **pandas** | Fast to write, rich API, great for prototyping |
| Joins across huge tables, production queries | **SQL** | Optimized for relational operations at scale |
| Quick one-off analysis, non-technical stakeholders | **Excel** | Familiar, visual, easy to share |
| >10M rows, multi-core or distributed processing | **Dask** | Scales pandas workflows beyond memory limits |
| Large datasets, performance-critical analytics | **Polars** | Fast, modern, columnar execution engine |
| GPU-accelerated analytics | **cuDF** | Leverages NVIDIA GPUs for faster computation |

**Decision guide:** If it fits in memory and you are exploring, use pandas.

## 1.7 A Typical pandas Workflow

Every pandas analysis follows the same pattern: **load -> explore -> clean -> analyze -> export.**

Here is a preview using inline data. Each step is a one-liner — details come in later chapters.

````python
# 1. Load — create data inline (Chapter 4 covers file loading)
df = pd.DataFrame({
    'name': ['Alice', 'Bob', 'Charlie', 'Diana', 'Eve'],
    'department': ['Sales', 'Engineering', 'Sales', 'HR', 'Engineering'],
    'salary': [50000, 75000, 55000, None, 68000]
})
print("Step 1 - Load:")
print(df)
print()
````

````python
# 2. Explore — what does the data look like? (Chapter 2)
print("Step 2 - Explore:")
print(f"Shape: {df.shape}")
print(f"Types:\n{df.dtypes}")
print()
````

````python
# 3. Clean — handle the missing value (Chapter 10)
df_clean = df.dropna()
print("Step 3 - Clean (drop rows with missing values):")
print(f"Rows before: {len(df)}, Rows after: {len(df_clean)}")
print()
````

````python
# 4. Analyze — average salary by department (Chapter 8)
result = df_clean.groupby('department')['salary'].mean()
print("Step 4 - Analyze (mean salary by department):")
print(result)
print()
````

````python
# 5. Export — save the result (Chapter 4)
result.to_csv('department_salaries.csv')
print("Step 5 - Export: saved to department_salaries.csv")
````

**What to notice:** Diana (the only HR employee) had a missing salary and was dropped in step 3. So the result has no HR row. This is exactly the kind of subtle behavior that trips people up — and that this book teaches you to catch.

### Solution 2: Series vs DataFrame

````python
employees = pd.DataFrame({
    'name': ['Alice', 'Bob', 'Charlie', 'Diana'],
    'department': ['Sales', 'Engineering', 'Sales', 'HR'],
    'salary': [50000, 75000, 55000, 60000]
})

# Single brackets -> Series (1D)
salary_series = employees['salary']
print("Single brackets (Series):")
print(salary_series)
print(f"Type: {type(salary_series)}")
print()

# Double brackets -> DataFrame (2D)
salary_df = employees[['salary']]
print("Double brackets (DataFrame):")
print(salary_df)
print(f"Type: {type(salary_df)}")
````

**What you should see:** Single brackets return a `pandas.core.series.Series` — a flat list of values with an index. Double brackets return a `pandas.core.frame.DataFrame` — a table with one column. Same data, different structure. The double brackets matter.

### Solution 4: pandas vs Plain Python

````python
employees_raw = [
    {'name': 'Alice', 'salary': 70000},
    {'name': 'Bob', 'salary': 85000},
    {'name': 'Charlie', 'salary': 60000},
    {'name': 'Diana', 'salary': 92000},
]

# Approach 1: Plain Python loop (3 lines)
total = 0
for emp in employees_raw:
    total += emp['salary']
avg_python = total / len(employees_raw)
print(f"Python loop: {avg_python}")

# Approach 2: pandas one-liner (1 line after DataFrame creation)
avg_pandas = pd.DataFrame(employees_raw)['salary'].mean()
print(f"pandas:      {avg_pandas}")

print(f"\nSame result: {avg_python == avg_pandas}")
print("The pandas version is one line and reads like English:")
print("'take salary column, compute mean.'")
````

**What you should see:** Both approaches return `76750.0`. The Python loop took 3-4 lines of manual work. The pandas version is a single expression: `pd.DataFrame(employees_raw)['salary'].mean()`. At 4 rows the difference is small — at 200,000 rows the pandas version is also much faster.
