# Core pandas Data Structures: Series and DataFrame

## Introduction

pandas provides two fundamental data structures that form the foundation of data manipulation in Python: **Series** and **DataFrame**. Understanding when and how to use each is crucial for effective data analysis work.

Think of these structures as specialized containers designed to handle labeled data with semantic meaning. Unlike NumPy arrays where axes are largely interchangeable, pandas structures assign meaningful labels to rows and columns, making your code more readable and less error-prone.

---

## Series: One-Dimensional Labeled Arrays

### What is a Series?

A **Series** is a one-dimensional labeled array capable of holding any data type. Think of it as a single column in a spreadsheet—each value has an associated label called an **index**. It is the fundamental building block of pandas.

### Creating a Series

The simplest way to create a Series is from a Python list. When you don't supply an index, pandas assigns a default integer index starting at zero:

````python
import pandas as pd
import numpy as np

# From a list — default integer index
temperatures = pd.Series([72, 75, 68, 71, 73])
print(temperatures)
````

Output:
```
0    72
1    75
2    68
3    71
4    73
dtype: int64
```

The real power of Series emerges when you assign meaningful labels:

````python
# From a list with a custom index
temperatures = pd.Series(
    [72, 75, 68, 71, 73],
    index=['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday'],
    name='Daily Temperature (°F)'
)
print(temperatures)
print(f"\nTemperature on Wednesday: {temperatures['Wednesday']}")
````

Output:
```
Monday       72
Tuesday      75
Wednesday    68
Thursday     71
Friday       73
Name: Daily Temperature (°F), dtype: int64

Temperature on Wednesday: 68
```

You can also create a Series from other sources:

````python
# From a dictionary — keys become the index
stock_prices = pd.Series({
    'AAPL': 150.25,
    'GOOGL': 2800.50,
    'MSFT': 310.75
})
print(stock_prices)

# From a scalar value — broadcasts to fill the index
constant_series = pd.Series(5, index=['A', 'B', 'C', 'D'])
print(constant_series)

# From a NumPy array with a custom index
quarterly = pd.Series(np.array([10, 20, 30, 40]), index=['Q1', 'Q2', 'Q3', 'Q4'])
print(quarterly)
````

### Key Properties of a Series

````python
temps = pd.Series(
    [72, 75, 68, 71, 73],
    index=['Mon', 'Tue', 'Wed', 'Thu', 'Fri'],
    name='Daily Temperature (°F)'
)

print(temps.index)    # Index(['Mon', 'Tue', 'Wed', 'Thu', 'Fri'], dtype='object')
print(temps.values)   # [72 75 68 71 73]
print(temps.dtype)    # int64
print(temps.name)     # Daily Temperature (°F)
print(f"Shape: {temps.shape}")   # (5,)
print(f"Size:  {temps.size}")    # 5
````

### Accessing Data in a Series

pandas provides several ways to retrieve values. Prefer label-based access (`.loc` / direct label) for clarity, and position-based access (`.iloc`) when you need numeric offsets:

````python
temps = pd.Series([72, 75, 68, 71, 73],
                  index=['Mon', 'Tue', 'Wed', 'Thu', 'Fri'])

# Label-based access
print(temps['Wed'])                    # 68
print(temps[['Mon', 'Wed', 'Fri']])   # Multiple labels

# Position-based access
print(temps.iloc[2])      # 68
print(temps.iloc[1:3])    # Tue and Wed

# Fast single-value accessors
print(temps.at['Mon'])    # Label-based — fastest for one value
print(temps.iat[0])       # Position-based — fastest for one value

# Boolean indexing
print(temps[temps > 72])  # Days warmer than 72°F
````

### Operations on a Series

Arithmetic and functions operate element-wise, so you rarely need explicit loops:

````python
# Arithmetic
celsius = (temps - 32) * 5 / 9
print(celsius)

# Vectorized operations
doubled = temps * 2
print(doubled)

# Applying a custom function
converted = temps.apply(lambda x: round((x - 32) * 5 / 9, 1))
print(converted)
````

### Iterating Over a Series

Iteration is occasionally useful for side effects (e.g., printing), but prefer vectorized operations for performance:

````python
# items() yields (label, value) pairs
for day, temp in temps.items():
    print(f"{day}: {temp}°F")

# Direct iteration yields values only
for temp in temps:
    print(temp)
````

## DataFrame: Two-Dimensional Labeled Tables

### What is a DataFrame?

A **DataFrame** is a two-dimensional labeled structure with columns that can each hold a different data type. Think of it as a spreadsheet or SQL table—it has rows, columns, and cells. It is the most commonly used pandas data structure.

### Creating a DataFrame

The most common approach is to pass a dictionary of lists, where each key becomes a column name:

````python
# From a dictionary of lists
data = {
    'Name':       ['Alice', 'Bob', 'Charlie', 'Diana'],
    'Age':        [25, 30, 35, 28],
    'Salary':     [50000, 60000, 75000, 55000],
    'Department': ['Sales', 'IT', 'Finance', 'HR']
}
df = pd.DataFrame(data)
print(df)
````

Output:
```
      Name  Age  Salary Department
0    Alice   25   50000      Sales
1      Bob   30   60000         IT
2  Charlie   35   75000    Finance
3    Diana   28   55000         HR
```

You can supply a meaningful row index at creation time:

````python
df = pd.DataFrame(data, index=['EMP001', 'EMP002', 'EMP003', 'EMP004'])
print("Column names:", df.columns.tolist())
print("Row indices: ", df.index.tolist())
print("Shape:       ", df.shape)   # (4, 4)
````

Other common creation patterns:

````python
# From a list of dictionaries (each dict is one row)
records = [
    {'product': 'Laptop',   'price': 1200, 'stock': 5},
    {'product': 'Mouse',    'price': 25,   'stock': 50},
    {'product': 'Keyboard', 'price': 75,   'stock': 30}
]
df_records = pd.DataFrame(records)

# From a NumPy array
arr = np.random.randn(3, 4)
df_array = pd.DataFrame(arr,
                        columns=['A', 'B', 'C', 'D'],
                        index=['Row1', 'Row2', 'Row3'])

# From a dictionary of Series (index alignment happens automatically)
sales_df = pd.DataFrame({
    'Q1': pd.Series([100, 200, 300], index=['Product A', 'Product B', 'Product C']),
    'Q2': pd.Series([150, 250, 320], index=['Product A', 'Product B', 'Product C'])
})
print(sales_df)
````

### Key Properties of a DataFrame

````python
df = pd.DataFrame({
    'Name':   ['Alice', 'Bob', 'Charlie'],
    'Age':    [25, 30, 35],
    'Salary': [50000, 60000, 75000]
})

print(df.shape)    # (3, 3) — rows, columns
print(df.size)     # 9 — total cells
print(df.columns)  # Index(['Name', 'Age', 'Salary'], dtype='object')
print(df.index)    # RangeIndex(start=0, stop=3, step=1)
print(df.dtypes)   # Name: object, Age: int64, Salary: int64
print(df.describe())
print(df.info())
````

### Accessing Data in a DataFrame

````python
df = pd.DataFrame({
    'Name':   ['Alice', 'Bob', 'Charlie'],
    'Age':    [25, 30, 35],
    'Salary': [50000, 60000, 75000]
}, index=['EMP001', 'EMP002', 'EMP003'])

# Single column → returns a Series
print(df['Name'])
print(type(df['Name']))   # <class 'pandas.core.series.Series'>

# Multiple columns → returns a DataFrame
print(df[['Name', 'Salary']])

# Row by position
print(df.iloc[0])         # First row as a Series

# Row by label
print(df.loc['EMP001'])   # Row with index 'EMP001'

# Specific cell
print(df.loc['EMP001', 'Salary'])   # 50000
print(df.at['EMP001', 'Salary'])    # 50000 — faster for a single value

# Boolean indexing
high_earners = df[df['Salary'] > 55000]
print(high_earners)
````

### Working with Columns

````python
# Add a new column
df['Bonus'] = df['Salary'] * 0.1

# Modify an existing column
df['Age'] = df['Age'] + 1

# Drop a column (returns a new DataFrame by default)
df_dropped = df.drop('Bonus', axis=1)

# Rename columns
df_renamed = df.rename(columns={'Salary': 'Annual_Salary'})
````

### Iterating Over a DataFrame

Iteration is useful for inspection, but vectorized operations are always faster for computation:

````python
# Iterate over column names
for col in df.columns:
    print(f"Column: {col}")

# Iterate over rows as Series — convenient but slow on large data
for idx, row in df.iterrows():
    print(f"Row {idx}: {row['Name']}")

# Iterate over rows as named tuples — faster than iterrows()
for row in df.itertuples():
    print(row)

# Iterate over (column_name, Series) pairs
for col_name, col_data in df.items():
    if col_data.dtype != object:
        print(f"{col_name} mean: {col_data.mean():.1f}")
````

## The Relationship Between Series and DataFrame

A DataFrame is essentially a dictionary of Series objects, where each column is a Series sharing the same index:

```
DataFrame (2D container)
    ├── Column "Name"   → Series of strings
    ├── Column "Age"    → Series of integers
    └── Column "Salary" → Series of integers
```

````python
df = pd.DataFrame({
    'Temperature': [72, 68, 75],
    'Humidity':    [45, 50, 48]
})

# Accessing a column returns a Series
temp_series = df['Temperature']
print(type(temp_series))   # <class 'pandas.core.series.Series'>

# Accessing a row also returns a Series
row_series = df.iloc[0]
print(type(row_series))    # <class 'pandas.core.series.Series'>

# You can build a DataFrame from individual Series
s1 = pd.Series([1, 2, 3], name='Column1')
s2 = pd.Series([4, 5, 6], name='Column2')
combined_df = pd.DataFrame({'Col1': s1, 'Col2': s2})
print(combined_df)
````

This relationship means that operations you learn on Series apply directly when you extract a column from a DataFrame.

## Mutability and Copying

Both structures are **value-mutable** (you can change cell values), but they differ in **size-mutability**:

````python
# Most operations return new objects — the original is unchanged
original = pd.Series([1, 2, 3, 4, 5])
doubled = original * 2

print(original)  # [1, 2, 3, 4, 5]
print(doubled)   # [2, 4, 6, 8, 10]

# Series length is fixed; to extend, create a new Series
# pd.concat() is the correct way to combine Series
extended = pd.concat([original, pd.Series([6, 7])])
print(extended)

# DataFrames are more flexible — columns can be added
df = pd.DataFrame({'A': [1, 2, 3], 'B': [4, 5, 6]})
df['C'] = [7, 8, 9]          # Add a column in-place
df_dropped = df.drop('B', axis=1)  # Returns a new DataFrame

# Use .copy() to make a fully independent copy
df_copy = df.copy()
df_copy['D'] = [10, 11, 12]
print(df)       # Original — no column D
print(df_copy)  # Copy — has column D
````

## Series vs. DataFrame: Decision Guide

| Aspect | Series | DataFrame |
|--------|--------|-----------|
| **Dimensions** | 1D | 2D |
| **Data types** | Typically single dtype (can be mixed via `object`) | Column-wise dtypes (each column can differ) |
| **Use case** | Single variable or column | Multiple related variables |
| **Indexing** | Single index | Row index + column labels |
| **Common source** | Single column, computed results | CSV files, databases, tabular data |
| **Structure flexibility** | Fixed 1D structure | Columns can be added or removed |
| **Analogy** | One column in a spreadsheet | The entire spreadsheet |

### When to Use Series

- **Single variable analysis**: Temperature readings, stock prices over time
- **Intermediate calculations**: Extracting a column from a DataFrame for processing
- **Function return values**: Many pandas operations return a Series

````python
# Analyzing daily returns for a single stock
daily_returns = pd.Series(
    [0.02, -0.01, 0.03, 0.01, -0.02],
    index=pd.date_range('2024-01-01', periods=5),
    name='AAPL Returns'
)

cumulative = (1 + daily_returns).cumprod() - 1
print(cumulative)
````

### When to Use DataFrame

- **Multi-variable datasets**: Multiple related columns describing the same entities
- **Tabular data**: Data naturally organized in rows and columns
- **Data from files**: Reading CSV, Excel, or database files
- **Complex analysis**: Filtering, grouping, and transforming multiple columns simultaneously

````python
# Analyzing portfolio performance across multiple stocks
portfolio = pd.DataFrame({
    'Date':  pd.date_range('2024-01-01', periods=5),
    'AAPL':  [150, 152, 151, 153, 155],
    'GOOGL': [140, 141, 140, 142, 143],
    'MSFT':  [380, 382, 381, 383, 385]
})

# Calculate daily returns across all stocks at once
returns = portfolio[['AAPL', 'GOOGL', 'MSFT']].pct_change()
print(returns)
````

## Practical Example: Choosing the Right Structure

The following scenario shows both structures working together on the same data:

````python
import pandas as pd

# --- Series: one student's grades across subjects ---
alice_grades = pd.Series(
    [85, 92, 78, 88],
    index=['Math', 'English', 'Science', 'History'],
    name='Alice'
)
print("Alice's grades (Series):")
print(alice_grades)
print(f"Average: {alice_grades.mean()}\n")

# --- DataFrame: the whole class ---
grades_df = pd.DataFrame({
    'Alice':   [85, 92, 78, 88],
    'Bob':     [90, 85, 92, 80],
    'Charlie': [78, 88, 85, 90]
}, index=['Math', 'English', 'Science', 'History'])

print("Class grades (DataFrame):")
print(grades_df)

print("\nAverage by subject:")
print(grades_df.mean(axis=1))

print("\nAverage by student:")
print(grades_df.mean(axis=0))

# Extracting Alice's column from the DataFrame gives the same Series
alice_from_df = grades_df['Alice']
print(f"\nSame data, extracted from DataFrame:\n{alice_from_df}")
````

---

## Summary

- **Series**: Use for single-dimensional, labeled data—one column with an index.
- **DataFrame**: Use for two-dimensional, tabular data—multiple columns sharing the same row index.
- Both structures provide semantic meaning through labels, making analysis code more readable and less error-prone.
- A DataFrame is composed of Series; extracting a column always returns a Series, so skills transfer directly between the two.
- Many pandas operations return new objects rather than modifying originals, which promotes safer and more predictable data manipulation.
- Series are 1D labeled structures with a fixed index. To combine multiple Series, use operations like `pd.concat()`. DataFrames allow columns to be added and removed dynamically.

Understanding these core structures and when to use each is the foundation for everything else in pandas.

````python
import pandas as pd
import numpy as np

original = pd.DataFrame({
    'item': ['Laptop', 'Mouse', 'Keyboard', 'Monitor'],
    'price': [999.0, 25.0, 75.0, 349.0],
    'in_stock': [True, True, False, True]
})

# Create a variable 'ref' that points to the SAME DataFrame (no copy)
ref = original

# Update the price of 'Mouse' (row index 1) to 29.99 via 'ref'
ref.at[1, 'price'] = 29.99

# Print the 'price' column of 'original' — it will have changed too
print("Original prices after modifying ref:")
print(original['price'])

# Create a TRUE independent copy of 'original'
discounted = original.copy()

# Apply a 10% discount to all prices in 'discounted'
discounted['price'] = discounted['price'] * 0.9

# Print both price columns to confirm 'original' is unchanged
print("\nOriginal prices (should be unchanged):")
print(original['price'])
print("\nDiscounted prices:")
print(discounted['price'])
````
