# Applying Functions and Method Chaining

Pandas lets you apply custom functions to your data for powerful, flexible transformations. Combined with method chaining, you can build clear, efficient workflows.

## Prerequisites: Understanding Lambda Functions

Before diving into `.apply()`, let's understand **lambda functions** — anonymous functions perfect for simple, one-time operations:

````python
import pandas as pd
import numpy as np

# Lambda syntax: lambda arguments: expression

# Example 1: Simple lambda
square = lambda x: x ** 2
print(square(5))  # Output: 25

# Example 2: Lambda with multiple arguments
add = lambda x, y: x + y
print(add(3, 4))  # Output: 7

# Lambda functions are most useful when passed to pandas methods
# They're concise and don't require a separate function definition
````

**Why use lambda?** When you need a simple, one-time function, lambda is faster to write than defining a full function with `def`.

## Applying Functions

### 1. Column-wise Operations (axis=0, default)

Apply a function to **each value in a column** independently:

````python
import pandas as pd

# Create sample data
df = pd.DataFrame({
    'age': [15, 25, 45, 70, 35],
    'salary': [0, 45000, 75000, 50000, 60000],
    'name': ['Alice', 'Bob', 'Charlie', 'Diana', 'Eve']
})

print("Original DataFrame:")
print(df)
print()

# Method 1: Using a custom function
def age_group(age):
    """Categorize age into groups"""
    if age < 18:
        return 'Child'
    elif age < 65:
        return 'Adult'
    else:
        return 'Senior'

# Apply function to each value in 'age' column
df['age_group'] = df['age'].apply(age_group)

print("After applying age_group function:")
print(df)
print()

# Method 2: Using lambda (more concise for simple operations)
df['salary_bracket'] = df['salary'].apply(lambda x: 'High' if x > 60000 else 'Low')

print("After applying lambda for salary brackets:")
print(df)
````

### 2. Row-wise Operations (axis=1)

Apply a function **across columns for each row**:

````python
import pandas as pd

# Create sample data with multiple score columns
df = pd.DataFrame({
    'name': ['Alice', 'Bob', 'Charlie'],
    'math': [85, 92, 78],
    'english': [90, 88, 95],
    'science': [88, 85, 82]
})

print("Original DataFrame:")
print(df)
print()

# Calculate total score across all subjects for each student
# axis=1 means apply function to each row
df['total_score'] = df[['math', 'english', 'science']].apply(
    lambda row: row.sum(),
    axis=1
)

print("After calculating total_score (row-wise):")
print(df)
print()

# Calculate average score for each student
df['average_score'] = df[['math', 'english', 'science']].apply(
    lambda row: row.mean(),
    axis=1
)

print("After calculating average_score:")
print(df)
````

- `axis=1`: Apply to each **row** independently

### 3. Using `.map()` for Fast Value Lookups

`.map()` is **Series-only** and optimized for simple dictionary lookups. It's **faster than `.apply()`** for this use case:

````python
import pandas as pd

# Create sample data
df = pd.DataFrame({
    'student': ['Alice', 'Bob', 'Charlie', 'Diana'],
    'grade': ['A', 'B', 'A', 'C']
})

print("Original DataFrame:")
print(df)
print()

# Method 1: Using .map() with dictionary (FAST for lookups)
grade_to_gpa = {'A': 4.0, 'B': 3.0, 'C': 2.0, 'D': 1.0, 'F': 0.0}
df['gpa'] = df['grade'].map(grade_to_gpa)

print("After mapping grades to GPA:")
print(df)
print()

# Method 2: .map() with a function (also works)
df['grade_points'] = df['grade'].map(lambda x: ord(x) - 64)  # A=1, B=2, C=3, etc.

print("After mapping grades to points:")
print(df)
print()

# Method 3: .map() with default value for missing keys
grade_to_level = {'A': 'Excellent', 'B': 'Good'}
df['level'] = df['grade'].map(grade_to_level, na_action='ignore')
# Note: Missing grades (C, D, F) will be NaN

print("After mapping with missing values:")
print(df)
````

**`.apply()` vs `.map()` Performance Comparison:**

````python
import pandas as pd
import numpy as np
import time

# Create large Series
large_series = pd.Series(np.random.choice(['A', 'B', 'C'], 100000))
grade_map = {'A': 4.0, 'B': 3.0, 'C': 2.0}

# Time .map()
start = time.time()
result_map = large_series.map(grade_map)
time_map = time.time() - start

# Time .apply()
start = time.time()
result_apply = large_series.apply(lambda x: grade_map[x])
time_apply = time.time() - start

print(f".map() time: {time_map:.4f}s")
print(f".apply() time: {time_apply:.4f}s")
print(f".map() is {time_apply/time_map:.1f}x faster")
````

**When to use `.map()`:**
- Simple dictionary lookups
- Mapping values to new values
- Need speed for large Series

### 4. Using `.transform()` for Group-wise Operations

`.transform()` applies a function **within each group** and returns a result with the **same shape** as the original data:

````python
import pandas as pd

# Create sample data with departments
df = pd.DataFrame({
    'department': ['Sales', 'Sales', 'IT', 'IT', 'HR', 'HR'],
    'employee': ['Alice', 'Bob', 'Charlie', 'Diana', 'Eve', 'Frank'],
    'salary': [50000, 55000, 75000, 80000, 45000, 48000]
})

print("Original DataFrame:")
print(df)
print()

# Calculate average salary per department
print("Average salary by department:")
print(df.groupby('department')['salary'].mean())
print()

# Use .transform() to add average salary as a new column
# This returns a Series with the same index as original data
df['dept_avg_salary'] = df.groupby('department')['salary'].transform('mean')

print("After adding department average salary:")
print(df)
print()

# Normalize salaries within each department (z-score)
df['salary_normalized'] = df.groupby('department')['salary'].transform(
    lambda x: (x - x.mean()) / x.std()
)

print("After normalizing salaries within departments:")
print(df)
print()

# Calculate salary rank within each department
df['dept_rank'] = df.groupby('department')['salary'].transform('rank')

print("After ranking salaries within departments:")
print(df)
````

### 5. Error Handling in Functions

What happens if your function fails on some values? Use try-except:

````python
import pandas as pd
import numpy as np

# Create data with problematic values
df = pd.DataFrame({
    'id': [1, 2, 3, 4, 5],
    'divisor': [2, 0, 4, 0, 5]  # Some zeros will cause division errors
})

print("Original DataFrame:")
print(df)
print()

# Function WITH error handling (safe)
def safe_divide(x):
    """Safely divide 100 by x, returning None if x is 0"""
    try:
        return 100 / x
    except ZeroDivisionError:
        return None

df['result'] = df['divisor'].apply(safe_divide)

print("After applying safe_divide:")
print(df)
print()

# Alternative: Use conditional lambda
df['result_coerce'] = df['divisor'].apply(
    lambda x: 100 / x if x != 0 else np.nan
)

print("Using conditional lambda:")
print(df)
````

## Performance: Vectorized Operations vs `.apply()`

**`.apply()` can be slow for large datasets.** Use vectorized operations when possible:

````python
import pandas as pd
import numpy as np
import time

# Create large dataset
df = pd.DataFrame({
    'age': np.random.randint(0, 100, 100000),
    'value': np.random.randn(100000)
})

print("DataFrame shape:", df.shape)
print()

# Method 1: Using .apply() with lambda (SLOW)
start = time.time()
result_apply = df['age'].apply(
    lambda x: 'Child' if x < 18 else 'Adult' if x < 65 else 'Senior'
)
time_apply = time.time() - start

# Method 2: Using pd.cut() (FAST - vectorized)
start = time.time()
result_cut = pd.cut(
    df['age'],
    bins=[0, 18, 65, 101],  # 101 ensures age 100 is included
    labels=['Child', 'Adult', 'Senior'],
    right=False             # [0,18), [18,65), [65,101)
)
time_cut = time.time() - start

# Method 3: Using np.select() (FAST - vectorized)
start = time.time()
conditions = [
    df['age'] < 18,
    (df['age'] >= 18) & (df['age'] < 65),
    df['age'] >= 65
]
choices = ['Child', 'Adult', 'Senior']

result_select = np.select(
    conditions,
    choices,
    default='Unknown'  # Must be a string to match choices dtype
)
time_select = time.time() - start

print(f"Method 1 (.apply):  {time_apply:.6f}s")
print(f"Method 2 (pd.cut): {time_cut:.6f}s - {time_apply/time_cut:.1f}x faster")
print(f"Method 3 (np.select): {time_select:.6f}s - {time_apply/time_select:.1f}x faster")

# Verify results are identical
print("\nResults match:")
print("apply vs cut:", (result_apply == result_cut.astype(str)).all())
print("apply vs select:", (result_apply == result_select).all())
````

**Example Output:**
```
DataFrame shape: (100000,)

Method 1 (.apply): 0.1234s
Method 2 (pd.cut): 0.0045s - 27.4x faster
Method 3 (np.select): 0.0032s - 38.5x faster
```

**When to use vectorized operations instead of `.apply()`:**
- Conditional logic: Use `pd.cut()`, `np.where()`, or `np.select()`
- String operations: Use the `.str` accessor
- Date operations: Use the `.dt` accessor
- Mathematical operations: Use NumPy or pandas built-in functions

## Method Chaining

Chain multiple operations for concise, readable code. Each method returns an object that the next method can operate on:

### Basic Chaining Example

````python
import pandas as pd

# Create sample data
df = pd.DataFrame({
    'name': ['Alice', 'Bob', 'Charlie', 'Diana', 'Eve'],
    'age': [25, 30, 35, 28, 32],
    'salary': [50000, 60000, 75000, 55000, 65000],
    'department': ['Sales', 'IT', 'IT', 'HR', 'Sales']
})

print("Original DataFrame:")
print(df)
print()

# Method 1: Without chaining (multiple intermediate variables)
df_filtered = df[df['salary'] > 55000]
df_with_group = df_filtered.assign(
    salary_level=lambda x: x['salary'].apply(lambda s: 'High' if s > 65000 else 'Medium')
)
result = df_with_group.groupby('department')['salary'].mean()

print("Result without chaining:")
print(result)
print()

# Method 2: With chaining (cleaner, more readable)
result = (
    df
    .query('salary > 55000')
    .assign(salary_level=lambda x:
            x['salary'].apply(lambda s: 'High' if s > 65000 else 'Medium'))
    .groupby('department')['salary']
    .mean()
)

print("Result with chaining:")
print(result)
````

**Output:**
```

      name  age  salary department
0    Alice   25   50000      Sales
1      Bob   30   60000        IT
2  Charlie   35   75000        IT
3    Diana   28   55000        HR
4      Eve   32   65000      Sales

Result without chaining:
Result with chaining:
department
IT       75000.0
Sales    65000.0
Name: salary, dtype: float64
```

### Advanced Chaining with Multiple Transformations

````python
import pandas as pd

# Create sample data with missing values
df = pd.DataFrame({
    'date': pd.date_range('2024-01-01', periods=5),
    'product': ['A', 'B', 'A', 'C', 'B'],
    'quantity': [10, 20, None, 15, 25],
    'price': [100, 150, 200, None, 175]
})

print("Original DataFrame:")
print(df)
print()

# Complex chain: Clean → Transform → Aggregate
result = (
    df
    .dropna()                          # Remove rows with missing values
    .assign(
        revenue=lambda x: x['quantity'] * x['price'],
        month=lambda x: x['date'].dt.month
    )
    .groupby('product')                # Group by product
    .agg({
        'revenue': 'sum',
        'quantity': 'mean',
        'price': 'max'
    })
    .round(2)                          # Round to 2 decimal places
)

print("After chaining:")
print(result)
````

**Output:**
```

1 2024-01-02       B      20.0    150

3 2024-01-04       C      15.0    NaN
4 2024-01-05       B      25.0    175

After chaining:
         revenue  quantity  price
product
```

### Chaining Best Practices

````python
import pandas as pd
import numpy as np

# Create sample data
df = pd.DataFrame({
    'age': [25, 28, 35, 42, 50, 23, 38, 45, 29, 55],
    'salary': [50000, 55000, 65000, 75000, 85000, 48000, 70000, 80000, 52000, 90000]
})

# Clear, readable chaining with proper formatting
result = (
    df
    .dropna(subset=['salary'])
    .query('age >= 25')
    .assign(
        age_group=lambda x: pd.cut(x['age'], bins=[0, 30, 40, 100],
                                    labels=['Young', 'Mid', 'Senior']),
        bonus=lambda x: x['salary'] * 0.1
    )
    .groupby('age_group', observed=False)
    .agg({'salary': 'mean', 'bonus': 'sum'})
    .round(2)
)

print(result)
````

**Output:**
```
           salary    bonus
age_group
Young      51000.0  10300.0
Mid        70000.0  14000.0
Senior     87500.0  17500.0
```

## Common Patterns & Best Practices

### Pattern 1: Conditional Column Creation

````python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    'score': [45, 78, 92, 55, 88]
})

# Avoid: Multiple nested .apply() calls
# df['grade'] = df['score'].apply(
#     lambda x: 'A' if x >= 90 else 'B' if x >= 80 else 'C' if x >= 70 else 'F'
# )

# Better: Use pd.cut() for ranges
df['grade_cut'] = pd.cut(df['score'],
                         bins=[0, 70, 80, 90, 100],
                         labels=['F', 'C', 'B', 'A'])

# Better: Use np.select() for complex conditions
df['grade'] = np.select(
    [df['score'] >= 90, df['score'] >= 80, df['score'] >= 70],
    ['A', 'B', 'C'],
    default='F'
)

print(df)
````

**Output:**
```
   score grade_cut grade
0     45         F     F
1     78         C     C
2     92         A     A
3     55         F     F
4     88         B     B
```

### Pattern 2: String Operations

````python
import pandas as pd

df = pd.DataFrame({
    'name': ['alice smith', 'bob jones', 'charlie brown'],
    'email': ['alice@example.com', 'bob@example.com', 'charlie@example.com']
})

# Avoid: Using .apply() for string operations
# df['name_upper'] = df['name'].apply(lambda x: x.upper())

# Better: Use .str accessor (vectorized)
df['name_upper'] = df['name'].str.upper()
df['first_name'] = df['name'].str.split().str[0]
df['domain'] = df['email'].str.extract(r'@(.+)')[0]

print(df)
````

**Output:**
```
           name                email    name_upper first_name     domain
0   alice smith  alice@example.com  ALICE SMITH      alice  example.com
1     bob jones    bob@example.com    BOB JONES        bob  example.com
2 charlie brown charlie@example.com CHARLIE BROWN    charlie  example.com
```

### Pattern 3: Date Operations

````python
import pandas as pd

df = pd.DataFrame({
    'date': pd.date_range('2024-01-01', periods=5),
    'value': [10, 20, 30, 40, 50]
})

# Avoid: Using .apply() for date operations
# df['month'] = df['date'].apply(lambda x: x.month)

# Better: Use .dt accessor (vectorized)
df['month'] = df['date'].dt.month
df['year'] = df['date'].dt.year
df['day_name'] = df['date'].dt.day_name()

print(df)
````

**Output:**
```
        date  value  month  year day_name

```

### Exercise 1: Apply Functions

````python
import pandas as pd

# Create a DataFrame with student data
students = pd.DataFrame({
    'name': ['Alice', 'Bob', 'Charlie', 'Diana'],
    'math': [85, 92, 78, 88],
    'english': [90, 88, 95, 92],
    'science': [88, 85, 82, 90]
})

# 1. Create a function that returns 'Pass' if average score >= 80, else 'Fail'
def check_status(row):
    average = row[['math', 'english', 'science']].mean()
    return 'Pass' if average >= 80 else 'Fail'

# 2. Apply it to create a 'status' column
students['status'] = students.apply(check_status, axis=1)

# 3. Create a 'total_score' column using axis=1
students['total_score'] = students[['math', 'english', 'science']].sum(axis=1)

# 4. Create a 'best_subject' column showing which subject had the highest score
students['best_subject'] = students[['math', 'english', 'science']].idxmax(axis=1)

print(students)
````

### Exercise 2: Map and Transform

````python
import pandas as pd

sales = pd.DataFrame({
    'region': ['North', 'South', 'North', 'East', 'South', 'East'],
    'sales': [50000, 60000, 55000, 75000, 65000, 70000]
})

# 1. Create a mapping: North→'Region A', South→'Region B', East→'Region C'
region_mapping = {
    'North': 'Region A',
    'South': 'Region B',
    'East': 'Region C'
}

# 2. Use .map() to create a 'region_code' column
sales['region_code'] = sales['region'].map(region_mapping)

# 3. Use .transform() to add 'region_avg_sales' showing average sales per region
sales['region_avg_sales'] = sales.groupby('region')['sales'].transform('mean')

# 4. Calculate 'sales_vs_avg' showing difference from region average
sales['sales_vs_avg'] = sales['sales'] - sales['region_avg_sales']

print(sales)
````

### Exercise 3: Method Chaining

````python
import pandas as pd

employees = pd.DataFrame({
    'name': ['Alice', 'Bob', 'Charlie', 'Diana', 'Eve'],
    'department': ['Sales', 'IT', 'IT', 'HR', 'Sales'],
    'salary': [50000, 75000, 80000, 55000, 60000],
    'years': [2, 5, 7, 3, 4]
})

# Create a chain that:
# 1. Filters for employees with more than 3 years of experience
# 2. Adds a 'bonus' column equal to 10% of salary
# 3. Groups by department and calculates mean salary and total bonus
result = (
    employees
    .query('years > 3')
    .assign(bonus=lambda x: x['salary'] * 0.1)
    .groupby('department')
    .agg({'salary': 'mean', 'bonus': 'sum'})
    .round(2)
)

print(result)
````

````python
import pandas as pd
import numpy as np

# Sample temperature data
df = pd.DataFrame({
    'city': ['New York', 'London', 'Tokyo', 'Sydney', 'Paris'],
    'temp_celsius': [22.0, 15.5, 30.1, 18.3, 12.8]
})

# Create 'temp_f_vectorized' column using a vectorized formula
df['temp_f_vectorized'] = (df['temp_celsius'] * 9 / 5) + 32

# Create 'temp_f_apply' column using .apply() with a lambda
df['temp_f_apply'] = df['temp_celsius'].apply(
    lambda x: (x * 9 / 5) + 32
)

# Verify both columns are equal using .equals()
columns_match = df['temp_f_vectorized'].equals(df['temp_f_apply'])
print('Columns match:', columns_match)
print(df)
````

### Solution 3: Applying a Custom Function Across Rows

````python
import pandas as pd
import numpy as np

# Student scores data
df = pd.DataFrame({
    'student': ['Ana', 'Ben', 'Cara', 'Dan', 'Ella'],
    'homework': [85, 90, 78, 92, 88],
    'midterm':  [70, 85, 80, 75, 95],
    'final':    [80, 88, 74, 85, 91]
})

# Define a function that computes a weighted grade from a row
def weighted_grade(row):
    return (row['homework'] * 0.20) + (row['midterm'] * 0.30) + (row['final'] * 0.50)

# Apply the function row-wise to create a 'final_grade' column
df['final_grade'] = df.apply(weighted_grade, axis=1)

# Round 'final_grade' to 2 decimal places
df['final_grade'] = df['final_grade'].round(2)

print(df)
````

### Solution 4: Method Chaining Pipeline

````python
import pandas as pd
import numpy as np

# Raw orders data
df = pd.DataFrame({
    'product':  ['Widget', 'Gadget', 'Widget', 'Doohickey', 'Gadget', 'Widget', 'Doohickey'],
    'quantity': [5, 2, 8, 1, 4, 3, 6],
    'unit_price': [9.99, 24.99, 9.99, 49.99, 24.99, 9.99, 49.99],
    'returned': [False, False, True, False, False, False, True]
})

# Method chaining pipeline
result = (
    df
    .loc[df['returned'] == False]                          # Step 1: filter out returned orders
    .assign(total_price=lambda x: x['quantity'] * x['unit_price'])  # Step 2: add total_price
    .loc[lambda x: x['total_price'] > 50]                 # Step 3: filter total_price > 50
    .sort_values('total_price', ascending=False)           # Step 4: sort descending
    .reset_index(drop=True)                                # Step 5: reset index
)

print(result)
````
