# Indexing and Selecting Data: loc, iloc, Boolean Masks, and Index Behavior

Data selection is one of the most fundamental skills in pandas. Whether you're filtering rows by condition, extracting specific columns, or navigating complex index structures, understanding the different indexing methods will make your data analysis workflows significantly more efficient and readable.

## Understanding pandas Indexing Methods

pandas provides several complementary approaches to selecting data, each suited to different scenarios:

- **`loc`**: Label-based selection using index and column names
- **`iloc`**: Position-based selection using integer locations
- **`at` / `iat`**: Fast access to single values
- **Boolean masks**: Conditional filtering across rows or columns

Choosing the right method depends on whether you're working with meaningful labels or positional indices, and whether you need single values or multiple rows/columns.

## Label-Based Selection with `loc`

The `loc` accessor selects data using the actual index labels and column names. This is the most intuitive method when your data has meaningful row and column identifiers.

### Selecting Single Values

```python
import pandas as pd
import numpy as np

# Create a sample DataFrame with meaningful index
sales = pd.DataFrame({
    'product': ['Widget A', 'Widget B', 'Widget C'],
    'q1_revenue': [15000, 22000, 18500],
    'q2_revenue': [16200, 21800, 19100],
    'q3_revenue': [17500, 23100, 20300]
}, index=['north', 'south', 'east'])

# Access a single value by row and column label
print(sales.loc['south', 'q2_revenue'])  # Output: 21800
```

### Selecting Rows by Label

```python
# Select a single row
print(sales.loc['north'])
# product          Widget A
# q1_revenue          15000
# q2_revenue          16200
# q3_revenue          17500

# Select multiple rows
print(sales.loc[['north', 'east']])
#        product  q1_revenue  q2_revenue  q3_revenue
# north  Widget A       15000       16200       17500
# east   Widget C       18500       19100       20300
```

### Selecting Columns by Label

```python
# Select a single column (returns a Series)
print(sales.loc[:, 'product'])
# north    Widget A
# south    Widget B
# east     Widget C

# Select multiple columns
print(sales.loc[:, ['product', 'q1_revenue', 'q3_revenue']])
```

### Slicing with Labels

An important distinction from Python lists: label-based slicing is **inclusive on both ends**.

```python
# Slice from 'north' to 'east' (both included)
print(sales.loc['north':'east', 'q1_revenue':'q2_revenue'])
#        q1_revenue  q2_revenue
# north       15000       16200
# south       22000       21800
# east        18500       19100
```

## Position-Based Selection with `iloc`

The `iloc` accessor uses integer positions (0, 1, 2, ...) rather than labels. This is useful when you need to select data by its location, or when working with data that doesn't have meaningful index labels.

### Selecting by Integer Position

```python
# Select the first row
print(sales.iloc[0])
# product          Widget A
# q1_revenue          15000
# q2_revenue          16200
# q3_revenue          17500

# Select rows at positions 0 and 2
print(sales.iloc[[0, 2]])
#        product  q1_revenue  q2_revenue  q3_revenue
# north  Widget A       15000       16200       17500
# east   Widget C       18500       19100       20300
```

### Slicing with Integer Positions

Unlike label-based slicing, integer slicing is **exclusive on the right end** (standard Python behavior).

```python
# Get rows from position 0 up to (but not including) position 2
print(sales.iloc[0:2])
#        product  q1_revenue  q2_revenue  q3_revenue
# north  Widget A       15000       16200       17500
# south  Widget B       22000       21800       23100

# Get columns from position 1 to 3 (exclusive on right)
print(sales.iloc[:, 1:3])
#        q1_revenue  q2_revenue
# north       15000       16200
# south       22000       21800
# east        18500       19100
```

### Selecting Both Axes with `iloc`

```python
# Get specific rows and columns by position
print(sales.iloc[0:2, 1:3])  # Rows 0-1, columns 1-2
#        q1_revenue  q2_revenue
# north       15000       16200
# south       22000       21800

# Single value by position
print(sales.iloc[1, 2])  # Row 1, column 2
# Output: 21800
```

## Fast Single-Value Access with `at` and `iat`

When you need to access a single value, `at` and `iat` are optimized for speed and provide cleaner syntax than `loc` or `iloc`.

```python
# Using 'at' for label-based access
value = sales.at['south', 'q2_revenue']
print(value)  # Output: 21800

# Using 'iat' for position-based access
value = sales.iat[1, 2]
print(value)  # Output: 21800
```

These accessors are particularly useful in loops or when performing many single-value lookups, as they're faster than their broader counterparts.

## Understanding Index Behavior

The index is the "backbone" of a pandas DataFrame. Understanding how it works is crucial for effective data selection.

### Default Integer Index

When you create a DataFrame without specifying an index, pandas automatically creates a default integer index starting from 0.

```python
df = pd.DataFrame({
    'name': ['Alice', 'Bob', 'Charlie'],
    'age': [25, 30, 35]
})
print(df.index)  # RangeIndex(start=0, stop=3, step=1)
```

### Custom Index

You can set a meaningful index when creating a DataFrame or afterward:

```python
# Set index during creation
df = pd.DataFrame({
    'name': ['Alice', 'Bob', 'Charlie'],
    'age': [25, 30, 35]
}, index=['emp001', 'emp002', 'emp003'])

# Or set it afterward
df.index = ['emp001', 'emp002', 'emp003']

# Or use set_index() to promote a column to index
df = df.set_index('name')
```

### Index Alignment

One of pandas' most powerful features is automatic index alignment. When you perform operations on DataFrames or Series, pandas aligns them by index:

```python
series1 = pd.Series([10, 20, 30], index=['A', 'B', 'C'])
series2 = pd.Series([1, 2, 3], index=['B', 'C', 'D'])

# Addition aligns by index; missing indices become NaN
result = series1 + series2
print(result)
# A     NaN
# B    21.0
# C    32.0
# D     NaN
```

### Checking Index Membership

```python
# Check if a label exists in the index
print('south' in sales.index)  # True
print('west' in sales.index)   # False

# Get the position of a label
position = sales.index.get_loc('south')
print(position)  # 1
```

## Boolean Mask Filtering

Boolean masks allow you to filter data based on conditions. A boolean mask is an array of `True` and `False` values with the same length as your data, where `True` indicates rows to keep.

### Creating and Applying Boolean Masks

```python
# Create a boolean mask for rows where q1_revenue > 16000
mask = sales['q1_revenue'] > 16000
print(mask)
# north    False
# south     True
# east     False

# Apply the mask to get filtered rows
print(sales[mask])
#        product  q1_revenue  q2_revenue  q3_revenue
# south  Widget B       22000       21800       23100
```

### Combining Multiple Conditions

Use `&` (and), `|` (or), and `~` (not) operators to combine conditions. Always use parentheses around each condition.

```python
# Rows where q1_revenue > 16000 AND q3_revenue < 24000
mask = (sales['q1_revenue'] > 16000) & (sales['q3_revenue'] < 24000)
print(sales[mask])
#        product  q1_revenue  q2_revenue  q3_revenue
# south  Widget B       22000       21800       23100

# Rows where product contains 'Widget A' OR q2_revenue > 22000
mask = (sales['product'] == 'Widget A') | (sales['q2_revenue'] > 22000)
print(sales[mask])
#        product  q1_revenue  q2_revenue  q3_revenue
# north  Widget A       15000       16200       17500
# south  Widget B       22000       21800       23100

# Rows where q1_revenue is NOT > 20000
mask = ~(sales['q1_revenue'] > 20000)
print(sales[mask])
#        product  q1_revenue  q2_revenue  q3_revenue
# north  Widget A       15000       16200       17500
# east   Widget C       18500       19100       20300
```

### Why `&`, `|`, `~` Instead of `and`, `or`, `not`?

Python's `and`/`or` operators work on single boolean values and cannot evaluate an entire Series element-by-element. Using them with a pandas Series raises a `ValueError`. The bitwise operators `&`, `|`, and `~` work element-wise on each row.

```python
# ❌ WRONG - This raises a ValueError
# result = sales[(sales['q1_revenue'] > 16000) and (sales['q3_revenue'] < 24000)]

# ✅ CORRECT - Use & instead
result = sales[(sales['q1_revenue'] > 16000) & (sales['q3_revenue'] < 24000)]
```

### Using `isin()` for Multiple Values

```python
# Select rows where product is either 'Widget A' or 'Widget C'
mask = sales['product'].isin(['Widget A', 'Widget C'])
print(sales[mask])
#        product  q1_revenue  q2_revenue  q3_revenue
# north  Widget A       15000       16200       17500
# east   Widget C       18500       19100       20300
```

### Boolean Masks with `loc`

You can pass boolean masks directly to `loc` for more complex filtering:

```python
# Filter rows and select specific columns using a mask
mask = sales['q1_revenue'] > 16000
print(sales.loc[mask, ['product', 'q3_revenue']])
#        product  q3_revenue
# south  Widget B       23100
```

## Practical Workflow: Combining Selection Methods

Real-world data analysis often requires combining multiple selection techniques:

```python
# Scenario: Find the average Q2 revenue for regions
# where Q1 revenue exceeded $16,000

# Step 1: Create a boolean mask
high_performers = sales['q1_revenue'] > 16000

# Step 2: Filter using the mask
filtered_sales = sales[high_performers]

# Step 3: Select specific column and calculate
avg_q2 = filtered_sales.loc[:, 'q2_revenue'].mean()
print(f"Average Q2 revenue for high performers: ${avg_q2:,.0f}")
# Output: Average Q2 revenue for high performers: $21,800

# Or in one chained operation:
avg_q2 = sales[sales['q1_revenue'] > 16000]['q2_revenue'].mean()
```

## Common Pitfalls and Best Practices

### Pitfall: Confusing `loc` and `iloc` Slicing Behavior

```python
# loc: both endpoints included
print(sales.loc['north':'south'])  # Includes both 'north' AND 'south'

# iloc: right endpoint excluded (Python standard)
print(sales.iloc[0:2])  # Includes positions 0 and 1, NOT 2
```

### Pitfall: Forgetting Parentheses Around Conditions

Without parentheses, Python's operator precedence causes `&` to bind more tightly than `>`, producing incorrect results or errors.

```python
# ❌ WRONG - Operator precedence causes unexpected behavior
# result = sales[sales['q1_revenue'] > 16000 & sales['q3_revenue'] < 24000]

# ✅ CORRECT - Wrap each condition in parentheses
result = sales[(sales['q1_revenue'] > 16000) & (sales['q3_revenue'] < 24000)]
```

### Best Practice: Use `loc` for Label-Based Selection

When your data has meaningful labels, prefer `loc` over `iloc`. It makes your code more readable and less fragile to column reordering:

```python
# Good: Clear intent, works even if columns are reordered
revenue = sales.loc['south', 'q2_revenue']

# Less clear: Depends on column order
revenue = sales.iloc[1, 2]
```

### Best Practice: Use `.copy()` When Modifying Filtered Data

```python
# Create a copy to avoid SettingWithCopyWarning
filtered = sales[sales['q1_revenue'] > 16000].copy()
filtered['new_column'] = 100
```

### Best Practice: Use `.loc[]` for Conditional Assignment

When you want to modify values in the original DataFrame based on a condition, always use `.loc[]`:

```python
# ❌ RISKY - May not modify the original DataFrame
# sales[sales['q1_revenue'] > 16000]['q2_revenue'] = 99999

# ✅ CORRECT - Modifies the original DataFrame in place
sales.loc[sales['q1_revenue'] > 16000, 'q2_revenue'] = 99999
```

## Summary

| Method | Use Case | Example |
|--------|----------|---------|
| `loc` | Label-based selection | `df.loc['row_label', 'col_label']` |
| `iloc` | Position-based selection | `df.iloc[0, 2]` |
| `at` | Fast single value (label) | `df.at['row', 'col']` |
| `iat` | Fast single value (position) | `df.iat[0, 2]` |
| Boolean mask | Conditional filtering | `df[df['col'] > 100]` |

Mastering these selection techniques will dramatically improve your ability to manipulate and analyze data efficiently in pandas.

### Solution 1: Label-Based Selection with loc

```python
import pandas as pd
import numpy as np

data = {
    'name': ['Alice', 'Bob', 'Carol', 'David', 'Eve'],
    'department': ['HR', 'Engineering', 'Engineering', 'Marketing', 'HR'],
    'salary': [55000, 85000, 90000, 62000, 58000],
    'years': [3, 7, 5, 2, 4]
}
df = pd.DataFrame(data, index=['E001', 'E002', 'E003', 'E004', 'E005'])

# Use loc to select the row for employee 'E003'
row_e003 = df.loc['E003']
print('Row E003:')
print(row_e003)

# Use loc to select rows 'E001' through 'E004' (inclusive),
# but only the 'name' and 'salary' columns
subset = df.loc['E001':'E004', ['name', 'salary']]
print('\nE001 to E004, name and salary:')
print(subset)

# Use loc to select employees 'E002' and 'E005' (non-contiguous),
# returning all columns
selected = df.loc[['E002', 'E005']]
print('\nE002 and E005:')
print(selected)
```

### Solution 2: Position-Based Selection with iloc

```python
import pandas as pd
import numpy as np

data = {
    'product': ['Widget', 'Gadget', 'Doohickey', 'Thingamajig', 'Whatsit', 'Gizmo'],
    'category': ['A', 'B', 'A', 'C', 'B', 'C'],
    'price': [9.99, 24.99, 4.99, 49.99, 14.99, 34.99],
    'stock': [150, 80, 300, 45, 120, 60],
    'reorder_point': [50, 30, 100, 20, 40, 25]
}
df = pd.DataFrame(data)

# Use iloc to select the first 3 rows and first 3 columns
first_block = df.iloc[:3, :3]
print('First 3 rows, first 3 columns:')
print(first_block)

# Use iloc to select the last 2 rows (use negative indexing)
last_two = df.iloc[-2:]
print('\nLast 2 rows:')
print(last_two)

# Use iloc to select every other row (rows at positions 0, 2, 4)
# and only the columns at positions 0, 2, 3
every_other = df.iloc[::2, [0, 2, 3]]
print('\nEvery other row, columns product/price/stock:')
print(every_other)
```

### Solution 3: Boolean Mask Filtering

```python
import pandas as pd
import numpy as np

data = {
    'student': ['Ana', 'Ben', 'Cara', 'Dan', 'Ella', 'Finn', 'Gina', 'Hank'],
    'grade': [88, 72, 95, 61, 83, 91, 74, 55],
    'subject': ['Math', 'Science', 'Math', 'Science', 'Math', 'Science', 'Math', 'Science'],
    'passed': [True, True, True, False, True, True, True, False]
}
df = pd.DataFrame(data)

# Create a boolean mask for students with a grade above 80
# and print the filtered DataFrame
high_achievers = df['grade'] > 80
print('Students with grade > 80:')
print(df[high_achievers])

# Create a boolean mask for Math students who scored below 90
# (combine two conditions with &)
math_below_90 = (df['subject'] == 'Math') & (df['grade'] < 90)
print('\nMath students with grade < 90:')
print(df[math_below_90])

# Create a boolean mask for students who either failed (passed == False)
# OR scored above 90 (combine two conditions with |)
edge_cases = (df['passed'] == False) | (df['grade'] > 90)
print('\nStudents who failed OR scored above 90:')
print(df[edge_cases])

# Using the high_achievers mask, use .loc to return only
# the 'student' and 'grade' columns for those rows
high_names_grades = df.loc[high_achievers, ['student', 'grade']]
print('\nHigh achievers - name and grade only:')
print(high_names_grades)
```

### Solution 4: Combining Selection Methods in a Workflow

```python
import pandas as pd
import numpy as np

data = {
    'rep': ['Jordan', 'Morgan', 'Taylor', 'Casey', 'Riley', 'Avery'],
    'region': ['North', 'South', 'North', 'East', 'West', 'East'],
    'q1_sales': [12000, 8500, 15000, 9200, 11000, 7800],
    'q2_sales': [13500, 9100, 14200, 10500, 12300, 8900],
    'q3_sales': [11800, 9800, 16000, 9800, 10900, 9500]
}
df = pd.DataFrame(data, index=['R01', 'R02', 'R03', 'R04', 'R05', 'R06'])

# Add a new column 'total_sales' that is the sum of q1, q2, and q3 sales
df['total_sales'] = df['q1_sales'] + df['q2_sales'] + df['q3_sales']

# Use a boolean mask to find all reps in the 'North' or 'East' region
# (use the .isin() method on the 'region' column)
north_east_mask = df['region'].isin(['North', 'East'])
north_east = df[north_east_mask]
print('North and East reps:')
print(north_east)

# From the filtered north_east result, use iloc to get
# only the first 2 rows
top_two = north_east.iloc[:2]
print('\nFirst 2 North/East reps:')
print(top_two)

# Use .at to retrieve the single value of 'total_sales' for rep 'R03'
best_rep_total = df.at['R03', 'total_sales']
print(f'\nR03 total sales: {best_rep_total}')

# Use .iat to retrieve the value in row position 0, column position 3
# (which should be q2_sales for the first rep)
first_q2 = df.iat[0, 3]
print(f'First rep Q2 sales (via iat): {first_q2}')

# Use loc with a boolean mask to set 'q3_sales' to 0
# for any rep whose total_sales is below 28000
# Then print the updated DataFrame
df.loc[df['total_sales'] < 28000, 'q3_sales'] = 0
print('\nUpdated DataFrame (low performers q3 zeroed):')
print(df)
```
