### Strategy 2: Filling Missing ValuesWhen data loss is unacceptable, fill missing values strategically:

```python
# Forward fill: propagate last valid valuedf['status'] = df['status'].ffill()# Backward fill: use next valid valuedf['status'] = df['status'].bfill()# Fill with a specific valuedf['region'] = df['region'].fillna('Unknown')# Fill with statistical measuresdf['age'] = df['age'].fillna(df['age'].median())# Fill different columns with different strategiesfill_values = {    'age': df['age'].median(),    'income': df['income'].mean(),    'category': df['category'].mode()[0]}df = df.fillna(fill_values)# Fill with group-specific valuesdf['salary'] = df.groupby('department')['salary'].transform(    lambda x: x.fillna(x.mean()))
```

## A Complete Cleaning Workflow ExampleHere is how these techniques combine into a practical, reusable workflow:

```python
def clean_sales_data(filepath):    """    Complete cleaning workflow for sales data.    """    # 1. Load and explore    df = pd.read_csv(filepath)    print(f"Original shape: {df.shape}")    print(f"Missing values:\n{df.isnull().sum()}\n")    # 2. Remove duplicates    df = df.drop_duplicates(subset=['order_id'])    # 3. Fix data types    df['order_date'] = pd.to_datetime(df['order_date'])    df['amount'] = pd.to_numeric(df['amount'].str.replace('$', ''), errors='coerce')    df['quantity'] = pd.to_numeric(df['quantity'], errors='coerce')    # 4. Handle missing values    df['notes'] = df['notes'].fillna('No notes')    df['discount'] = df['discount'].fillna(0)    df['amount'] = df['amount'].fillna(df['amount'].median())    df['quantity'] = df['quantity'].fillna(1)    df = df.dropna(subset=['customer_id'])    # 5. Standardize strings    df['product_name'] = df['product_name'].str.strip().str.lower()    df['region'] = df['region'].str.upper().str.strip()    # 6. Handle outliers    Q1 = df['amount'].quantile(0.25)    Q3 = df['amount'].quantile(0.75)    IQR = Q3 - Q1    df = df[~((df['amount'] < Q1 - 1.5 * IQR) | (df['amount'] > Q3 + 1.5 * IQR))]    df = df[df['quantity'] > 0]    # 7. Validate and flag    df['is_valid'] = (        (df['amount'] > 0) &        (df['quantity'] > 0) &        (df['order_date'].notna())    )    invalid_records = df[~df['is_valid']]    if len(invalid_records) > 0:        print(f"Warning: {len(invalid_records)} invalid records found")    df = df[df['is_valid']].drop('is_valid', axis=1)    print(f"Final shape: {df.shape}")    return df# Usagedf_clean = clean_sales_data('sales_data.csv')
```

## Validation and Quality ChecksAfter cleaning, validate that your data meets requirements:

```python
def validate_cleaned_data(df):    """    Validate data quality after cleaning.    """    checks = {        'No missing values': df.isnull().sum().sum() == 0,        'No duplicates': df.duplicated().sum() == 0,        'Valid date range': (df['order_date'].min() >= pd.Timestamp('2020-01-01')),        'Positive amounts': (df['amount'] > 0).all(),        'Expected columns': all(col in df.columns for col in ['order_id', 'order_date', 'amount'])    }    for check_name, result in checks.items():        status = '✓' if result else '✗'        print(f"{status} {check_name}")    if all(checks.values()):        print("\n✓ All validation checks passed!")    else:        print("\n✗ Some validation checks failed. Review the issues above.")    return all(checks.values())is_valid = validate_cleaned_data(df_clean)
```

---### Recipe 6: Handling Missing Data by GroupReal-world data has missing values. Filling them globally can distort analysis, but filling within groups preserves patterns. This is essential for multi-category or multi-time-series datasets.

```python
import pandas as pdimport numpy as npprint("RECIPE 6: HANDLING MISSING DATA BY GROUP\n")print("="*50 + "\n")# Create sample data with missing valuesdf = pd.DataFrame({    'group': ['A', 'A', 'A', 'A', 'B', 'B', 'B', 'B'],    'date':  pd.date_range('2024-01-01', periods=4).tolist() * 2,    'value': [1.0, np.nan, 3.0, 4.0, 10.0, 11.0, np.nan, 13.0]})print("Original data with missing values:")print(df)print("\n" + "="*50 + "\n")# Recipe 6a: Forward fill within groupsprint("Recipe 6a: Forward fill within groups:")print("(Fill missing with previous value in same group)")df['value_ffill'] = df.groupby('group')['value'].ffill()print(df)print("\nExplanation:")print("  Group A: NaN at row 1 → filled with 1.0 (from row 0)")print("  Group B: NaN at row 6 → filled with 11.0 (from row 5)")print("\n" + "="*50 + "\n")# Recipe 6b: Backward fill within groupsprint("Recipe 6b: Backward fill within groups:")print("(Fill missing with next value in same group)")df['value_bfill'] = df.groupby('group')['value'].bfill()print(df[['group', 'date', 'value', 'value_bfill']])print("\nExplanation:")print("  Group A: NaN at row 1 → filled with 3.0 (from row 2)")print("  Group B: NaN at row 6 → filled with 13.0 (from row 7)")print("\n" + "="*50 + "\n")# Recipe 6c: Fill with group meanprint("Recipe 6c: Fill with group mean:")df['value_mean'] = df.groupby('group')['value'].transform(    lambda x: x.fillna(x.mean()))print(df[['group', 'date', 'value', 'value_mean']])print("\nExplanation:")print("  Group A mean: (1 + 3 + 4) / 3 = 2.67")print("  Group B mean: (10 + 11 + 13) / 3 = 11.33")print("  NaN values filled with respective group means")print("\n" + "="*50 + "\n")# Recipe 6d: Fill with group medianprint("Recipe 6d: Fill with group median:")df['value_median'] = df.groupby('group')['value'].transform(    lambda x: x.fillna(x.median()))print(df[['group', 'date', 'value', 'value_median']])print("\n" + "="*50 + "\n")# Recipe 6e: Interpolate within groupsprint("Recipe 6e: Interpolate within groups:")print("(Fill missing with linear interpolation within each group)")df_sorted = df.sort_values(['group', 'date']).reset_index(drop=True)df_sorted['value_interp'] = df_sorted.groupby('group')['value'].transform(    lambda x: x.interpolate(method='linear'))print(df_sorted[['group', 'date', 'value', 'value_interp']])print("\nExplanation:")print("  Group A: NaN at row 1 → interpolated between 1.0 and 3.0 = 2.0")print("  Group B: NaN at row 6 → interpolated between 11.0 and 13.0 = 12.0")print("\n" + "="*50 + "\n")# Recipe 6f: Comparison tableprint("MISSING DATA FILL METHODS COMPARISON:")print("""┌──────────────┬──────────────┬──────────────────────────────┐│ Method       │ Use Case     │ Example Result               │├──────────────┼──────────────┼──────────────────────────────┤│ Forward Fill │ Time-series  │ [1, 1, 3, 4] (repeat last)   ││ (ffill)      │ data         │                              │├──────────────┼──────────────┼──────────────────────────────┤│ Backward     │ Time-series  │ [1, 3, 3, 4] (use next)      ││ Fill (bfill) │ data         │                              │├──────────────┼──────────────┼──────────────────────────────┤│ Group Mean   │ Cross-       │ [1, 2.67, 3, 4] (avg)        ││              │ sectional    │                              │├──────────────┼──────────────┼──────────────────────────────┤│ Group Median │ Skewed data  │ [1, 3, 3, 4] (median)        ││              │              │                              │├──────────────┼──────────────┼──────────────────────────────┤│ Interpolate  │ Smooth       │ [1, 2, 3, 4] (linear)        ││              │ sequences    │                              │└──────────────┴──────────────┴──────────────────────────────┘""")
```
