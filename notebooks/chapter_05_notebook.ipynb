# Basic DataFrame Operations

## Introduction

DataFrames are the core data structure in pandas, designed to handle tabular data with labeled rows and columns. This chapter covers the fundamental operations you'll use daily: selecting data, performing calculations, renaming columns, and applying transformations. These skills form the foundation for all more advanced data manipulation tasks.

## Selecting Columns

### Single Column Selection

The most common operation is extracting a single column. You can use bracket notation or dot notation:

```python
import pandas as pd

# Create a sample DataFrame
df = pd.DataFrame({
    'product': ['Laptop', 'Mouse', 'Keyboard', 'Monitor'],
    'price': [999, 25, 75, 299],
    'stock': [5, 150, 80, 12]
})

# Bracket notation (recommended for column names with spaces or special characters)
prices = df['price']

# Dot notation (cleaner but only works with valid Python identifiers)
products = df.product

print(type(prices))  # Returns: <class 'pandas.core.series.Series'>
```

Notice that selecting a single column returns a **Series**, not a DataFrame. This is important for understanding how pandas structures data.

### Multiple Column Selection

To select multiple columns and maintain a DataFrame structure, use a list of column names:

```python
# Select multiple columns
subset = df[['product', 'price']]

print(type(subset))  # Returns: <class 'pandas.core.frame.DataFrame'>
print(subset)

#       product  price
# 0      Laptop    999
# 1       Mouse     25
# 2    Keyboard     75
# 3     Monitor    299
```

The double brackets are crucial here—the outer brackets select from the DataFrame, and the inner brackets create a list of column names.

### Column Selection by Position

For position-based selection, use `.iloc[]`:

```python
# Select first two columns by position
first_two = df.iloc[:, 0:2]

# Select specific columns by index
columns_0_and_2 = df.iloc[:, [0, 2]]
```

## Basic Arithmetic Operations

### Element-wise Operations

Pandas makes arithmetic operations intuitive by applying them element-wise across all rows:

```python
# Multiply all prices by 1.1 (10% increase)
increased_prices = df['price'] * 1.1

# Create new column with calculated values
df['price_with_tax'] = df['price'] * 1.08

print(df[['price', 'price_with_tax']])

#   price  price_with_tax
# 0    999         1078.92
# 1     25           27.00
# 2     75           81.00
# 3    299          322.92
```

### Multi-column Arithmetic

Combine columns in calculations:

```python
# Calculate total inventory value
df['total_value'] = df['price'] * df['stock']

print(df[['product', 'price', 'stock', 'total_value']])

#       product  price  stock  total_value
# 0      Laptop    999      5         4995
# 1       Mouse     25    150         3750
# 2    Keyboard     75     80         6000
# 3     Monitor    299     12         3588
```

### Broadcasting with Scalars and Arrays

Pandas automatically broadcasts scalar values and arrays across all rows:

```python
import numpy as np

# Add a scalar to all values
df['price'] + 50

# Subtract an array (must match DataFrame length)
discounts = np.array([100, 5, 10, 50])
df['discounted_price'] = df['price'] - discounts

print(df[['price', 'discounted_price']])

#   price  discounted_price
# 0    999               899
# 1     25                20
# 2     75                65
# 3    299               249
```

## Renaming Columns

### Rename Specific Columns

Use the `.rename()` method to change column names. This returns a new DataFrame without modifying the original:

```python
# Rename specific columns
df_renamed = df.rename(columns={
    'product': 'item_name',
    'price': 'unit_price',
    'stock': 'quantity_available'
})

print(df_renamed.columns)

# Index(['item_name', 'unit_price', 'quantity_available', 'price_with_tax',
#        'total_value', 'discounted_price'], dtype='object')
```

### Rename All Columns

Apply a function to all column names:

```python
# Convert all column names to uppercase
df_upper = df.rename(columns=str.upper)

print(df_upper.columns)

# Index(['PRODUCT', 'PRICE', 'STOCK', 'PRICE_WITH_TAX', 'TOTAL_VALUE',
#        'DISCOUNTED_PRICE'], dtype='object')

# Add a prefix to all columns
df_prefixed = df.add_prefix('inventory_')

print(df_prefixed.columns)

# Index(['inventory_product', 'inventory_price', 'inventory_stock', ...], dtype='object')
```

### In-place Renaming

To modify the DataFrame directly without creating a new object:

```python
# Modify the original DataFrame
df.rename(columns={'product': 'item_name'}, inplace=True)
```

## Simple Transformations

### Type Conversions

Transform data types to suit your analysis:

```python
# Re-create the DataFrame so column names are consistent
df = pd.DataFrame({
    'product': ['Laptop', 'Mouse', 'Keyboard', 'Monitor'],
    'price': [999, 25, 75, 299],
    'stock': [5, 150, 80, 12]
})

# Convert to integer
df['stock'] = df['stock'].astype('int64')

# Convert to string
df['product'] = df['product'].astype('string')

# Convert a column to numeric (useful when reading from CSV)
df['price'] = pd.to_numeric(df['price'])

print(df.dtypes)

# product     string
# price      float64
# stock        int64
```

### String Operations

Apply string methods to text columns:

```python
# Convert to lowercase
df['product_lower'] = df['product'].str.lower()

# Get string length
df['product_length'] = df['product'].str.len()

# Replace text
df['product_clean'] = df['product'].str.replace('Laptop', 'Computer')

print(df[['product', 'product_lower', 'product_length']])

#       product product_lower  product_length
# 0      Laptop        laptop               6
# 1       Mouse         mouse               5
# 2    Keyboard      keyboard               8
# 3     Monitor       monitor               7
```

### Conditional Transformations

Create new columns based on conditions:

```python
import numpy as np
import pandas as pd

# Simple conditional using apply
df['price_category'] = df['price'].apply(
    lambda x: 'Expensive' if x > 500 else 'Affordable'
)

# Multiple conditions using np.select
conditions = [
    df['stock'] > 100,
    (df['stock'] > 20) & (df['stock'] <= 100),
    df['stock'] <= 20
]

choices = ['High Stock', 'Medium Stock', 'Low Stock']

df['stock_level'] = np.select(
    conditions,
    choices,
    default='Unknown'   # ✅ FIX: forces string dtype consistency
)

print(df[['product', 'stock', 'stock_level']])
```

### Using `.assign()` for Immutable Transformations

The `.assign()` method creates a new DataFrame with additional columns, useful for functional programming style:

```python
# Create new DataFrame with calculated columns
df_extended = df.assign(
    revenue=lambda x: x['price'] * x['stock'],
    profit_margin=lambda x: (x['price'] * 0.3) / x['price'] * 100
)

print(df_extended[['product', 'revenue', 'profit_margin']])

#       product  revenue  profit_margin
# 0      Laptop     4995           30.0
# 1       Mouse     3750           30.0
# 2    Keyboard     6000           30.0
# 3     Monitor     3588           30.0
```

Notice how `.assign()` returns a new DataFrame, leaving the original unchanged. This makes it ideal for chaining operations without side effects.

## Practical Example: Complete Workflow

Combining these operations into a realistic data processing pipeline:

```python
import pandas as pd

# Start with raw data
sales_data = pd.DataFrame({
    'item': ['Widget A', 'Widget B', 'Widget C'],
    'cost': [10, 15, 20],
    'units_sold': [100, 75, 150]
})

# Apply transformations using method chaining
result = (sales_data
    .rename(columns={'item': 'product_name', 'cost': 'unit_cost'})
    .assign(
        total_revenue=lambda x: x['unit_cost'] * x['units_sold'],
        profit_per_unit=lambda x: x['unit_cost'] * 0.4
    )
    .assign(
        total_profit=lambda x: x['profit_per_unit'] * x['units_sold']
    )
)

print(result)

#   product_name  unit_cost  units_sold  total_revenue  profit_per_unit  total_profit
# 0     Widget A         10         100           1000              4.0           400
# 1     Widget B         15          75           1125              6.0           450
# 2     Widget C         20         150           3000              8.0          1200
```

Each step in the chain returns a new DataFrame, so the original `sales_data` is never modified. This pattern—rename, then assign computed columns—is a common and readable approach to building data pipelines.

## Key Takeaways

- **Column selection** with single brackets returns a Series; double brackets return a DataFrame
- **Arithmetic operations** broadcast automatically across all rows
- **Renaming** with `.rename()` creates a new DataFrame by default; use `inplace=True` to modify in place
- **Transformations** can be applied using `.apply()`, string methods via `.str`, or `.assign()` for cleaner code
- **Chaining operations** with `.assign()` creates readable, functional-style data pipelines that leave the original DataFrame unchanged

These fundamental operations are building blocks for more complex data manipulation tasks you'll encounter in real-world analysis.
